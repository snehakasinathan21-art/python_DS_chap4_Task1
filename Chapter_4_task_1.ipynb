```python
#2.	Create a list of numbers and print only even numbers using a for loop
numbers=list(map(int,input().split()))
even_nos=[]
for i in numbers:
    if i%2==0:
        even_nos.append(i)
print(even_nos)
```

```python
#3.	Write a program that handles division by zero using try-except.
try:
    n = int(input("Enter a number:"))
    res= 100/n
except ZeroDivisionError:
    print("You can't divide by Zero")
except ValueError:
    print("Enter a valid number!")
else:
    print("Result is ",res)
finally:
    print("Execution complete")
```

```python
#4.	Given a string, count how many vowels it has
word = input("Enter a word").upper()
vowels =("A","E","I","O","U")
count=0
for char in word:
    if char in vowels:
        count+=1
print("No.of vowels in word are ",count)
        

```

```python
#9.	Replace all missing values in a DataFrame with the column’s mean.
import pandas as pd
import numpy as np
student_marks={"Marks": [80, 90, np.nan, 70],
    "Age": [20, np.nan, 22, 21]}
updated_marks=pd.DataFrame(student_marks)
new_marks=updated_marks.fillna(updated_marks.mean())
new_marks

```
